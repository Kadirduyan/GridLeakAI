# GridLeak AI
## Model Değerlendirme ve Karşılaştırma

Durum: Başlangıç iskeleti. Veri analizi ve model eğitimi henüz yapılmadı.
Hücreleri sırayla çalıştıracağız; gerçek bulgular işlemden sonra yazılacak.


### Değerlendirme protokolü

Kilitli test, pozitif sınıf ve karar eşiği tanımı.


### Başarı ölçüleri

Precision, recall, F1, accuracy, ROC-AUC ve average precision.


### Grafikler ve hata analizi

PR eğrileri, karmaşıklık matrisleri ve eğitim/CV farkları.


### Karşılaştırma ve Precision@K

Dört model ve dengeleme deneyleri; kapasiteye bağlı isabet.


### Sonuç ve öneriler

Avantaj/dezavantaj, seçim gerekçesi ve veri sınırları; sayılar deney sonrası.
